# Measurement Instrument for cyclomatic complexity

This measurement instrument extracts the data from one result of another measurement instrument - cccc

The input to this measurement instrument is NOT a measured entity, but a raw file. Therefore, this measurement instrument is a part of an entire chain of measurement instruments, hence part of the measurement method. 

## Part 1: Configuration

```python
import xml.etree.ElementTree as ET

# input file, which is a result from another measurement instrument
complexity_analysis_report_filename = './.cccc/anonymous.xml'
```

## Part 2: Finding the value

```python
complexity_analysis_report = ET.parse(complexity_analysis_report_filename)

document_root = complexity_analysis_report.getroot()
```

```python
cyclomatic_complexity_value = document_root.findall('./procedural_detail/member_function/McCabes_cyclomatic_complexity')

cyclomatic_complexity_number = cyclomatic_complexity_value[0].attrib['value']
```

## Part 3: assigning to base measure

```python
complexity = cyclomatic_complexity_number
```

## Important notes

I've designed this measurement instrument to specifically find the right value. This means that several things are hard-coded, for example:
* name of the measure: './procedural_detail/member_function/McCabes_cyclomatic_complexity'
* which measured entity I refer to: cyclomatic_complexity_value[0]

A good measurement instrument has these aspects configurable or even takes a list of all the values. 

However, this example is meant to illustrate the process of chaining measurement instruments into one measurement method.
